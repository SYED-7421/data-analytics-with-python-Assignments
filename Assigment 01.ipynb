```python
#TASK03



#Write a program in which use all the operators we can use in Python




# ---------- Variables ----------
a = 10
b = 3

# ---------- 1. Arithmetic Operators ----------
print("----- Arithmetic Operators -----")
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Exponentiation:", a ** b)

# ---------- 2. Comparison Operators ----------
print("\n----- Comparison Operators -----")
print("a == b:", a == b)
print("a != b:", a != b)
print("a > b:", a > b)
print("a < b:", a < b)
print("a >= b:", a >= b)
print("a <= b:", a <= b)

# ---------- 3. Logical Operators ----------
print("\n----- Logical Operators -----")
print("a > 5 and b < 5:", a > 5 and b < 5)
print("a > 5 or b > 5:", a > 5 or b > 5)
print("not(a > 5):", not(a > 5))

# ---------- 4. Assignment Operators ----------
print("\n----- Assignment Operators -----")
c = 5
print("c =", c)
c += 2
print("c += 2 ->", c)
c -= 1
print("c -= 1 ->", c)
c *= 3
print("c *= 3 ->", c)
c /= 2
print("c /= 2 ->", c)
c //= 2
print("c //= 2 ->", c)
c **= 2
print("c **= 2 ->", c)
c %= 5
print("c %= 5 ->", c)

# ---------- 5. Bitwise Operators ----------
print("\n----- Bitwise Operators -----")
print("a & b:", a & b)
print("a | b:", a | b)
print("a ^ b:", a ^ b)
print("~a:", ~a)
print("a << 2:", a << 2)
print("a >> 2:", a >> 2)

# ---------- 6. Identity Operators ----------
print("\n----- Identity Operators -----")
x = 10
y = 10
print("x is y:", x is y)
print("x is not y:", x is not y)

# ---------- 7. Membership Operators ----------
print("\n----- Membership Operators -----")
numbers = [1, 2, 3, 4, 5]
print("3 in numbers:", 3 in numbers)
print("10 not in numbers:", 10 not in numbers)
```
