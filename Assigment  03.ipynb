```python
n = int(input("Enter number of elements: "))
lst = []
for i in range(n):
    val = input(f"Enter element {i+1}: ")
    lst.append(val)

reversed_list = []
for i in range(len(lst) - 1, -1, -1):
    reversed_list.append(lst[i])

print("Original list:", lst)
print("Reversed list:", reversed_list)
```

```python
n = int(input("Enter number of elements: "))
lst = []
for i in range(n):
    val = int(input(f"Enter number {i+1}: "))
    lst.append(val)

largest = lst[0]
for num in lst:
    if num > largest:
        largest = num

print("Largest number is:", largest)
```

```python
n = int(input("Enter number of elements: "))
lst = []
for i in range(n):
    val = input(f"Enter element {i+1}: ")
    lst.append(val)

first_element = lst[0]
for i in range(len(lst) - 1):
    lst[i] = lst[i + 1]
lst[len(lst) - 1] = first_element

print("Rotated list:", lst)
```

Q5

```python
sentence = input("Enter a sentence: ")
word_to_delete = input("Enter the word to delete: ")

words = sentence.split()
new_words = []
for w in words:
    if w != word_to_delete:
        new_words.append(w)

new_sentence = " ".join(new_words)
print("Updated sentence:", new_sentence)
```

Q7

```python
def capitalize_words(sentence):
    words = sentence.split()
    new_words = []
    for w in words:
        new_word = w[0].upper() + w[1:]
        new_words.append(new_word)
    return " ".join(new_words)

text = input("Enter a sentence: ")
result = capitalize_words(text)
print("Output:", result)
```

```python
m = int(input("Enter number of rows: "))
n = int(input("Enter number of columns: "))

matrix = []
for i in range(m):
    row = []
    for j in range(n):
        val = int(input(f"Enter element [{i}][{j}]: "))
        row.append(val)
    matrix.append(row)

for i in range(m):
    row_sum = 0
    for j in range(n):
        row_sum += matrix[i][j]
    print(f"Sum of row {i+1} = {row_sum}")
```

Q9

```python
n = int(input("Enter number of rows: "))
m = int(input("Enter number of columns: "))

print("Enter first matrix:")
matrix1 = []
for i in range(n):
    row = []
    for j in range(m):
        val = int(input(f"Element [{i}][{j}]: "))
        row.append(val)
    matrix1.append(row)

print("Enter second matrix:")
matrix2 = []
for i in range(n):
    row = []
    for j in range(m):
        val = int(input(f"Element [{i}][{j}]: "))
        row.append(val)
    matrix2.append(row)

result = []
for i in range(n):
    row = []
    for j in range(m):
        row.append(matrix1[i][j] + matrix2[i][j])
    result.append(row)

print("Sum of matrices:")
for row in result:
    print(row)
```

Q10

```python
n1 = int(input("Enter rows of first matrix: "))
m1 = int(input("Enter columns of first matrix: "))
n2 = int(input("Enter rows of second matrix: "))
m2 = int(input("Enter columns of second matrix: "))

if m1 != n2:
    print("Matrix multiplication not possible!")
else:
    print("Enter first matrix:")
    matrix1 = []
    for i in range(n1):
        row = []
        for j in range(m1):
            val = int(input(f"Element [{i}][{j}]: "))
            row.append(val)
        matrix1.append(row)

    print("Enter second matrix:")
    matrix2 = []
    for i in range(n2):
        row = []
        for j in range(m2):
            val = int(input(f"Element [{i}][{j}]: "))
            row.append(val)
        matrix2.append(row)

    result = []
    for i in range(n1):
        row = []
        for j in range(m2):
            total = 0
            for k in range(m1):
                total += matrix1[i][k] * matrix2[k][j]
            row.append(total)
        result.append(row)

    print("Result of multiplication:")
    for row in result:
        print(row)
```
